```python
from transformers import AutoTokenizer
import argparse

tokenizer = AutoTokenizer.from_pretrained('Alpha-VLLM/Lumina-DiMOO', trust_remote_code=True)
```

```python

input_prompt = """
top-left / top-right / bottom-left / bottom-right of the image.
"""
print(f"Input Prompt:\n{input_prompt}\n")

```

```python
input_ids = tokenizer(input_prompt)['input_ids']
print(len(input_ids))
print(tokenizer.convert_ids_to_tokens(input_ids, skip_special_tokens=True))
```

```python
def generate_crop_size_list(num_patches, patch_size, max_ratio=4.0):
    assert max_ratio >= 1.0
    crop_size_list = []
    wp, hp = num_patches, 1
    while wp > 0:
        if max(wp, hp) / min(wp, hp) <= max_ratio:
            crop_size_list.append((wp * patch_size, hp * patch_size))
        if (hp + 1) * wp <= num_patches:
            hp += 1
        else:
            wp -= 1
    return crop_size_list
```

```python
print(generate_crop_size_list(1024, 32))
```
