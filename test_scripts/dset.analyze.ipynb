```python
from datasets import load_dataset

val_ds_stream = load_dataset("Jiwon-Kang/pixmo-count-filtered-imgContained", split="validation", streaming=True)

```

```python
val_ds_stream = val_ds_stream.filter(lambda count: count <= 10, input_columns=['count'])  # Example filter: keep only examples with count <= 10
```

```python
iter_ds = iter(val_ds_stream)
```

```python
from pprint import pprint
for _ in range(1):
    example = next(iter_ds)
    pprint(example)
```

```python
from PIL import Image
from io import BytesIO
img_bytes = example.get('image')
img_bytes = img_bytes['bytes']
image = Image.open(BytesIO(img_bytes)).convert("RGB")
image
```
